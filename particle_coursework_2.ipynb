# Question 1 (5 marks)
Unlike at LEP, where the polar angle of the scattered muons is measured with respect to the incoming electron-positron pair, we need to consider the directions and momenta of the quark-antiquark pair at a hadron collider such as the LHC. In general, as a result of different incoming quark and anti-quark momenta, the rest frame of the produced Z boson is not the same as the laboratory frame.

We can define the scattering angle $\cos\theta^*$ as the cosine of the polar angle of the negatively charged muon ($\mu^-$) in the vector boson’s rest frame (the Collins–Soper frame). This measures how “forward” ($\cos\theta^* > 0$) or “backward” ($\cos\theta^* < 0$) the lepton is emitted relative to the incoming quark direction.

- Give the expression for the angle $\cos\theta^*$ in the Collins-Soper frame and define all the quantities appearing in the formula.

$\cos\theta^* = \frac{2(P^+_1P^-_2 - P^-_1P^+_2)}{\sqrt{m^2_{ll}(m^2_{ll} + p^2_{T, ll}})}\frac{p_{z, ll}}{| p_{z, ll} |}$

Quantites:

$P^{\pm}_1 = (E_1 \pm p_{z, 1}) / \sqrt{2}$: light-cone momentum components of the negatively charged lepton. A normalised combination of $P_1$'s energy and longitudinal momentum (both expressed in GeV). Describes momentum along the beam (z-axis), causing the $P^+_1P^-_2$ and $P^-_1P^+_2$ products in the equation numerator to be invariant under a longitudinal Lorentz boost.

$P^{\pm}_2 = (E_2 \pm p_{z, 2}) / \sqrt{2}$: light-cone momentum components of the positively charged lepton. A normalised combination of $P_2$'s energy and longitudinal momentum (both expressed in GeV). Describes momentum along the beam (z-axis), causing the $P^+_1P^-_2$ and $P^-_1P^+_2$ products in the equation numerator to be invariant under a longitudinal Lorentz boost.

$m_{ll}$: invariant mass of the dilepton system (GeV/c$^2$).

$p_{T, ll}$: transverse momentum of the dilepton system (GeV).

$p_{z, ll}$: longitudinal momentum of the dilepton system (GeV).

As the given CMS paper states in the Introduction, this eqution gives the angle $\cos\theta^*$ in the Collins-Soper frame using laboratory frame quantities. The numerator measures assymetry between the leptons, and the denominator is the normalisation factor. The remaining term gives a sign factor, providing the direction of the longitudinal momentum. It is also noted the dilepton system in the current analysis refers to electrons and/or muons specifically. Here, natural units are used; $c = 1$.


# Question 2 (5 marks)

Experimentally, the forward-backward asymmetry $A_{FB}$ is given by the difference in the forward and backward cross-sections, divided by their sum: $A_{FB} = \frac{\sigma_{F} - \sigma_{B}}{\sigma_{F} + \sigma_{B}}$. This quantity is sensitive to the vector and axial-vector couplings of electroweak bosons to fermions. The forward and backward cross-sections are proportional to the number of events with leptons in the forward and backward direction (in the Collins-Soper frame).

- State here the equation that connects the forward-backward asymmetry as a function of the di-muon invariant mass, $A_{FB} (M)$, to the vector- and axial-vector couplings ($v_f$ and $a_f$) and write down their definitions. Also define all quantities that appear in this equation.

$A_{FB} (M) = a_la_q(8v_lv_q - Q_qKD_m) \times [16(v^2_l + a^2_l)(v^2_q + a^2_q) - 8v_lv_qQ_qKD_m + Q^2_qK^2(D^2_m + \Gamma^2_Z/m^2_Z)]^{-1}$

Quantities: (where $_q$, $_l$ refer to the participating quark or lepton, $_f$ to the fermion, and $_Z$ to the Z-boson.)

$a_f = T^f_3$: the axial-vector coupling of Z-bosons to fermions, equal to the 3rd component of the weak isospin of the fermion at leading order, where the above equation holds (dimensionless). Describes the strength of the axial-vector interaction between the Z-boson and the fermion.

$v_f = T^f_3 - 2Q_f\sin^2\theta_w$: the vector coupling of Z-bosons to fermions, where $Q_f$ represents the charge of a fermion in units of charge e, and $\sin^2\theta_w$ denotes the weak mixing angle (dimensionless). Describes the strength of the vector interaction between the Z-boson and the fermion.

The vector and axial-vector couplings describe the Z-boson interaction with fermions, whereas the photon couples only as a vector. The interference between photon and Z-boson exchange therefore contributes to the forward–backward asymmetry.

$K = 8\sqrt{2}\pi\alpha/G_Fm^2_Z$: where $\alpha$ is the electromagnetic coupling (1/137), $G_F$ the Fermi constant (1.17 $\times$ 10$^{-5}$ GeV$^{-2}$), and $m_Z$ is the mass of the Z-boson (91.2 GeV). A constant that determines the relative strengths of the electromagnetic and weak interactions in the equation by moderating the photon and Z-boson exchange contributions to the process (dimensionless).

$D_m = 1 - m^2_Z/m^2_{ll}$: the dilepton invariant mass relative to the Z-boson resonance. For example, $D_m$ = 0 when $m_Z$ = $M$ at the Z resonance. This leads to the strong dependence of $A_{FB}$ on dimuon mass (dimensionless).

$\Gamma_Z$: total decay width (here, of the Z-boson, GeV). Sum of partial decay widths of each possible decay channel. 

$M = m_{ll}$. The linked paper states this as equation (6) in the Introduction, also writing that this expression uses the quark direction provided by $\cos\theta^*$. In real experiments, quark direction is estimated using dilepton boost direction, introducing dilution into the asymmetry. Considering each term in the expression, the asymmetry varies with dilepton invariant mass, considering the fluctuating contributions by photon exchange, Z-boson exchange and their interference.

## Question 3 (10 marks)

The code example below shows how to construct a Lorentz vector from the energy and momentum components of a particle and how to add the transverse momentum, $p_T$, to the dataframe.

- Calculate the invariant mass of the muon pair, $M_{\mu^{+}\mu^{-}}$ as well as the angle $\cos\theta^*$ and add them to the dataframe.
- Plot both variables for all events in the dataframe.

```python

'''
Example using the vector object to extract arrays of Lorentz vectors for each muon, 
but you don't have to use this. More information here: 
https://vector.readthedocs.io/en/latest/src/object.html
in particular https://vector.readthedocs.io/en/latest/src/momentum4d.html
'''
mu1_lv = vector.array({"px":df["mu1_px"].to_numpy(), "py":df["mu1_py"].to_numpy(), "pz":df["mu1_pz"].to_numpy(),"E":df["mu1_E"].to_numpy()})
mu2_lv = vector.array({"px":df["mu2_px"].to_numpy(), "py":df["mu2_py"].to_numpy(), "pz":df["mu2_pz"].to_numpy(),"E":df["mu2_E"].to_numpy()})

df["pt1"] = mu1_lv.pt
df["pt2"] = mu2_lv.pt
```

```python
import numpy as np

# 1. Calculate invariant mass:

# Add the Lorentz vetors of the two muons.
mu1mu2_lv = mu1_lv + mu2_lv
# Calculate invariant mass using combined Lorentz vector sqrt(E^2 - px^2 - py^2 - pz^2) buit-in function.
df["mu1mu2_M"] = mu1mu2_lv.mass # add to df.

# 2. Calculate cos(theta*), using equation stated in Q1:

# Light-cone momentum components P+- 1, 2.
P1_plus = (df["mu1_E"] + df["mu1_pz"]) / np.sqrt(2)
P1_minus = (df["mu1_E"] - df["mu1_pz"]) / np.sqrt(2)
P2_plus = (df["mu2_E"] + df["mu2_pz"]) / np.sqrt(2)
P2_minus = (df["mu2_E"] - df["mu2_pz"]) / np.sqrt(2)

# Dimuon invariant mass, transverse and longitudinal momenta.
m12 = mu1mu2_lv.mass
pt12 = mu1mu2_lv.pt
sign_pz12 = np.sign(mu1mu2_lv.pz) # gives sign factor.

# Apply equation and add to df.
df["cos(theta*)"] = (2 * (P1_plus * P2_minus - P1_minus * P2_plus) / np.sqrt(m12**2 * (m12**2 + pt12**2))) * sign_pz12

# Sanity check: -1 < cos(theta*) < 1.
print(f"All cos(theta*) are between -1 & 1: {bool(df["cos(theta*)"].between(-1, 1).all())}.")

df
```

```python
import matplotlib.pyplot as plt

# Calculate histogram and statistical uncertainties using sqrt(# events in bins) for M and cos(theta*).
events_M, bin_edges_M = np.histogram(df["mu1mu2_M"], bins=100)
bin_centres_M = (bin_edges_M[:-1] + bin_edges_M[1:]) / 2
uncertainties_M = np.sqrt(events_M)

events_cos, bin_edges_cos = np.histogram(df["cos(theta*)"], bins=100, range=(-1, 1)) # including known range.
bin_centres_cos = (bin_edges_cos[:-1] + bin_edges_cos[1:]) / 2
uncertainties_cos = np.sqrt(events_cos)

fig, ax = plt.subplots(1, 2, figsize=(14, 6), layout="constrained")

# Plot mass histogram.
ax[0].stairs(events_M, bin_edges_M, label="Invariant mass")
# Add error bars from statistical uncertainties.
ax[0].errorbar(bin_centres_M,events_M,yerr=uncertainties_M,fmt="o",markersize=2,capsize=2,label="Statistical uncertainties")

ax[0].set_xlabel(r"Invariant mass $M_{\mu_1\mu_2}$ (GeV)")
ax[0].set_ylabel("Number of events")
ax[0].set_title("Invariant Mass Histogram")
ax[0].legend(loc="upper left", fontsize=12, frameon=True)

# Plot cos(theta*) histogram.
ax[1].stairs(events_cos, bin_edges_cos, label=r"$\cos\theta^*$")
# Add error bars from statistical uncertainties.
ax[1].errorbar(bin_centres_cos,events_cos,yerr=uncertainties_cos,fmt="o",markersize=2,capsize=2,label="Statistical uncertainties")

ax[1].set_xlabel(r"$\cos\theta^*$")
ax[1].set_ylabel("Number of events")
ax[1].set_title(r"$\cos\theta^*$ Histogram")
ax[1].legend(loc="lower left", fontsize=12, frameon=True)
```

## Question 4 (10 marks)

Now that we have calculated the angle $\cos\theta^*$, 

- plot the forward-backward asymmetry $A_{FB} = \frac{N_F - N_B}{N_F + N_B}$, where $N_F$ ($N_B$) are the number of events with $\cos\theta^* >0$ ($\cos\theta^* <0$), as a function of the di-muon invariant mass. 

Use the following range and bin boundaries in your histogram for the di-muon invariant mass (same as in CMS paper): 60,70,78,84,87,89,91,93,95,98,104,112,120.

```python
# Define given mass bin boundaries.
mass_bins = [60, 70, 78, 84, 87, 89, 91, 93, 95, 98, 104, 112, 120]

# Filter and sum cos(theta*) columns as a function of mass to find NF and NB as histograms.
N_F, _ = np.histogram(df["mu1mu2_M"][df["cos(theta*)"] > 0], bins=mass_bins)
N_B, _ = np.histogram(df["mu1mu2_M"][df["cos(theta*)"] < 0], bins=mass_bins) # _ placeholder as bin values already known.

# Caculate AFB using given equation.
A_FB = (N_F - N_B) / (N_F + N_B)

# Calculate bin centres.
mass_centres = (np.array(mass_bins[:-1]) + np.array(mass_bins[1:])) / 2

# Calculate statistical uncertainties.
# Partial differentiation error propagation used, assuming uncertainties of NF and NB are their square roots (Poisson distribution).
N_total = N_F + N_B
A_FB_error = np.sqrt((1 - A_FB**2) / N_total)

# Display histogram of results.
plt.figure(figsize=(10, 6))
plt.scatter(mass_centres,A_FB,color='red',s=50,label=r'$A_{FB}$')

# Plot asymmetry with statistical error bars.
plt.errorbar(mass_centres,A_FB,yerr=A_FB_error,fmt='none',capsize=3,elinewidth=2,label='Statistical uncertainty')

# Add horizontal zero line as in graphs in paper.
plt.axhline(y=0,c='black',linestyle='--')
plt.xlabel(r'$M_{\mu^+\mu^-}$ (GeV)')
plt.ylabel(r'$A_{FB}$')
plt.title('Forward-Backward Asymmetry vs Invariant Mass')

plt.xlim(60, 120)
plt.legend(loc='upper left', fontsize=15, frameon=True)
```

```python
# Displaying required values below:

asymmetry_df = pd.DataFrame({
    "Mass centre (GeV)": mass_centres,
    "Forward events": N_F,
    "Backward events": N_B,
    "A_FB": A_FB,
    "A_FB uncertainties": A_FB_error
})

print("ASYMMETRY VALUES:")
display(asymmetry_df)
```
